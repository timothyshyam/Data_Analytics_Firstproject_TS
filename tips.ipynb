```python
#TIP PERCENTAGE
tips["tip_percentage"] = (tips["tip"] / tips["total_bill"]) * 100 
```

```python
tips.head()
```

```python
tips["tip_percentage"].describe()
```

```python
#moving to the next requirement finding which day/time has the highest average tip%
avg_tip = tips.groupby(["day", "time"])["tip_percentage"].mean()

avg_tip
#it is not necessary that datashould be available for every possible day/time combination.
#For example, the original Tips dataset doesn't have Saturday lunch observations.
#That's completely fine. groupby() only shows combinations that actually exist in the dataset.
```

```python
avg_tip = tips.groupby(["day", "time"])["tip_percentage"].mean()

avg_tip = (
    avg_tip
    .sort_values(ascending=False)
    .to_frame(name="avg_tip")                                      
    .assign(record_count=tips.groupby(["day", "time"]).size()
    )
)

avg_tip
```

```python
# Friday Lunch has the highest average tip percentage at approximately 18.88%.
#However, Friday Lunch contains only 7 observations, so the result should be interpreted cautiously because of the relatively small sample size.
```
