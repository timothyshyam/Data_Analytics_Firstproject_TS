```python
import pandas as pd
from scipy.stats import ttest_ind

tips = pd.read_csv("tips_cleaned.csv")
```

```python
#Is there a significant difference in tip % between Smokers and Non-Smokers?
smokers = tips[tips["smoker"] == "Yes"]["tip_percentage"]
non_smokers = tips[tips["smoker"] == "No"]["tip_percentage"]
```

```python
print("Average tip % - Smokers:", smokers.mean())
print("Average tip % - Non-smokers:", non_smokers.mean())
```

```python
#perfroming t-test
t_stat, p_value = ttest_ind(
    smokers,
    non_smokers,
    equal_var=False
)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

```python
# Smokers had a slightly higher average tip percentage than non-smokers,
#However, the p-value is 0.682, which is greater than 0.05.
#Therefore, the test didn't find statistically significant evidence of a difference to say smoking status affects tip percentage.
```
